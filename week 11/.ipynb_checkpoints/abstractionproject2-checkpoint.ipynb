```python
from abc import ABC, abstractmethod
from tkinter import *
from tkinter import messagebox


# Abstract Base Class
class External_Vendors(ABC):

    @abstractmethod
    def menu(self):
        pass


# Cooperative Cafeteria Class
class CooperativeCafeteria(External_Vendors):

    def menu(self):
        items = """
COOPERATIVE CAFETERIA MENU

Jollof Rice and Stew  - ₦200
White Rice and Stew   - ₦200
Fried Rice            - ₦200
Salad                 - ₦100
Plantain              - ₦100
"""
        return items


# Faith Hostel Cafeteria Class
class FaithHostelCafeteria(External_Vendors):

    def menu(self):
        items = """
FAITH HOSTEL CAFETERIA MENU

Fried Rice            - ₦400
White Rice and Stew   - ₦400
Jollof Rice           - ₦400
Beans                 - ₦200
Chicken               - ₦1000
"""
        return items


# Student Centre Cafeteria Class
class StudentCentreCafeteria(External_Vendors):

    def menu(self):
        items = """
STUDENT CENTRE CAFETERIA MENU

Chicken Fried Rice    - ₦800
Pomo Sauce            - ₦300
Spaghetti Jollof      - ₦500
Amala/Ewedu           - ₦500
Semo with Eforiro Soup - ₦500
"""
        return items


# Function to display menu
def display_menu():

    vendor = vendor_entry.get().lower()

    if vendor == "cooperative":
        obj = CooperativeCafeteria()

    elif vendor == "faith hostel":
        obj = FaithHostelCafeteria()

    elif vendor == "student centre":
        obj = StudentCentreCafeteria()

    else:
        messagebox.showerror("Error", "Vendor not found!")
        return

    result_text.delete(1.0, END)
    result_text.insert(END, obj.menu())


# GUI Window
root = Tk()
root.title("PAU External Vendors")
root.geometry("500x450")
root.config(bg="lightblue")

# Heading
title = Label(root,
              text="PAU External Vendors Menu",
              font=("Arial", 18, "bold"),
              bg="lightblue")
title.pack(pady=10)

# Input Label
label = Label(root,
              text="Enter Vendor Name:",
              font=("Arial", 12),
              bg="lightblue")
label.pack()

# Entry Box
vendor_entry = Entry(root, width=30, font=("Arial", 12))
vendor_entry.pack(pady=5)

# Button
btn = Button(root,
             text="Display Menu",
             command=display_menu,
             bg="blue",
             fg="white",
             font=("Arial", 12, "bold"))
btn.pack(pady=10)

# Text Area
result_text = Text(root, width=50, height=15, font=("Courier New", 11))
result_text.pack(pady=10)

# Run Application
root.mainloop()
```
