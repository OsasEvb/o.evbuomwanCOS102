```python
from abc import ABC, abstractmethod


# Abstract Base Class
class coup_de_ecriva(ABC):

    @abstractmethod
    def Fan_Page(self):
        pass


# Subclasses
class FC_Cirok(coup_de_ecriva):
    def Fan_Page(self):
        print('Welcome to "FC Cirok"')


class Madiba_FC(coup_de_ecriva):
    def Fan_Page(self):
        print('Welcome to "Madiba FC"')


class Blue_Jay_FC(coup_de_ecriva):
    def Fan_Page(self):
        print('Welcome to "Blue Jay FC"')


class TSG_Walker(coup_de_ecriva):
    def Fan_Page(self):
        print('Welcome to "TSG Walker"')


# User input
club = input("Enter the club you support: ")

# Factory-style selection
if club.lower() == "fc cirok":
    obj = FC_Cirok()

elif club.lower() == "madiba fc":
    obj = Madiba_FC()

elif club.lower() == "blue jay fc":
    obj = Blue_Jay_FC()

elif club.lower() == "tsg walker":
    obj = TSG_Walker()

else:
    print("Club not found!")
    obj = None


# Method call
if obj:
    obj.Fan_Page()
```
