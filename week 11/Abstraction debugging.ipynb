```python
from abc import ABC, abstractmethod

# Abstract Class
class Product(ABC):

    def item_list(self, rate):
        print("amount submitted:", rate)

    @abstractmethod
    def product(self, rate):
        pass


# Child class
class Item(Product):

    def product(self, rate):
        print("Product rate is:", rate)


# Now instantiate the child class
r = Item()
r.product(3)
r.item_list(2)
```

```python
from abc import ABC, abstractmethod

class Geometric(ABC):

    @abstractmethod
    def volume(self):
        pass


class Rect(Geometric):
    def __init__(self, length, width, height):
        self.length = length
        self.width = width
        self.height = height

    def volume(self):
        return self.length * self.width * self.height


class Sphere(Geometric):
    def __init__(self, radius):
        self.radius = radius

    def volume(self):
        return (4/3) * 3.14 * self.radius ** 3


class Cube(Geometric):
    def __init__(self, edge):
        self.edge = edge

    def volume(self):
        return self.edge ** 3


# Example 3D shape (corrected idea)
class TriangularPrism(Geometric):
    def __init__(self, base, height, depth):
        self.base = base
        self.height = height
        self.depth = depth

    def volume(self):
        return 0.5 * self.base * self.height * self.depth


# Objects
rr = Rect(4, 6, 6)
ss = Sphere(8)
cc = Cube(5)
tt = TriangularPrism(5, 4, 3)

print("Volume of Rectangular Prism:", rr.volume())
print("Volume of Sphere:", ss.volume())
print("Volume of Cube:", cc.volume())
print("Volume of Triangular Prism:", tt.volume())
```

```python
from abc import ABC, abstractmethod

class Invoice(ABC):

    def final_bill(self, pay):
        print('Purchase of the product:', pay)

    @abstractmethod
    def invoice(self, pay):
        pass


class Paycheque(Invoice):
    def invoice(self, pay):
        print('Paycheque of:', pay)


class CardPayment(Invoice):
    def invoice(self, pay):
        print('Pay through card of:', pay)


# Usage
aa = Paycheque()
aa.invoice(6500)
aa.final_bill(6500)

print(isinstance(aa, Paycheque))
print(isinstance(aa, Invoice))

aa = CardPayment()
aa.invoice(2600)
aa.final_bill(2600)

print(isinstance(aa, CardPayment))
print(isinstance(aa, Invoice))
```

```python
from abc import ABC, abstractmethod

class Bank(ABC):
    def branch(self, Naira):
        print("Fees submitted : ",Naira)
   
    @abstractmethod
    def Bank(Naira):
        pass
    
class private(Bank):
    def Bank(Naira):
        print("Total Naira Value here: ",Naira)
        
class public(Bank):
    def Bank(Naira):
        print("Total Naira Value here:",Naira)

private.Bank(5000)
public.Bank(2000)

a = public()
a.branch(3500)
```

```python
from abc import ABC, abstractmethod

class Bank(ABC):

    def branch(self, Naira):
        print("Fees submitted:", Naira)

    @abstractmethod
    def bank(self, Naira):
        pass


class Private(Bank):
    def bank(self, Naira):
        print("Private bank total Naira value:", Naira)


class Public(Bank):
    def bank(self, Naira):
        print("Public bank total Naira value:", Naira)


# Objects
p = Private()
p.bank(5000)

q = Public()
q.bank(2000)

q.branch(3500)
```
