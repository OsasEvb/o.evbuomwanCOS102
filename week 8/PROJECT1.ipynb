```python
import tkinter as tk
from tkinter import messagebox

# Parent Class
class Zenith:
    def unique_services(self):
        return []

    def mutual_services(self):
        return ["Lines of credit", "Investment management and accounts", "Insurance"]


# Child Classes
class RetailBanking(Zenith):
    def unique_services(self):  # Polymorphism (override)
        return [
            "Retirement and education accounts",
            "Loans and mortgages",
            "Checking and savings"
        ]


class GlobalBanking(Zenith):
    def unique_services(self):  # Polymorphism (override)
        return [
            "Multi-currency management services",
            "Foreign currency accounts",
            "Foreign currency credit cards",
            "Transborder advisory services",
            "Liquidity management"
        ]


class CommercialBanking(Zenith):
    def unique_services(self):  # Polymorphism (override)
        return [
            "Advisory services"
        ]


# Function to display services
def show_services():
    name = entry_name.get()
    division = division_var.get()

    if not name:
        messagebox.showerror("Error", "Please enter employee name")
        return

    if division == "Retail":
        obj = RetailBanking()
    elif division == "Global":
        obj = GlobalBanking()
    elif division == "Commercial":
        obj = CommercialBanking()
    else:
        messagebox.showerror("Error", "Select a division")
        return

    services = obj.mutual_services() + obj.unique_services()

    result = f"Employee: {name}\nDivision: {division}\n\nServices:\n"
    result += "\n".join(f"- {s}" for s in services)

    text_output.delete(1.0, tk.END)
    text_output.insert(tk.END, result)


# GUI Setup
root = tk.Tk()
root.title("Zenith Bank Service Viewer")
root.geometry("400x400")

# Name input
tk.Label(root, text="Employee Name:").pack()
entry_name = tk.Entry(root)
entry_name.pack()

# Division selection
division_var = tk.StringVar()

tk.Label(root, text="Select Division:").pack()
tk.Radiobutton(root, text="Retail", variable=division_var, value="Retail").pack()
tk.Radiobutton(root, text="Global", variable=division_var, value="Global").pack()
tk.Radiobutton(root, text="Commercial", variable=division_var, value="Commercial").pack()

# Button
tk.Button(root, text="Show Services", command=show_services).pack(pady=10)

# Output box
text_output = tk.Text(root, height=10)
text_output.pack()

root.mainloop()
```
